### ICC模块

当数据存在多人标注的时候，为了保证特征的一致性，需要对特征进行ICC校验，这个模块需要在特征提取之后使用。例如
* Comp1-传统组学-传统组学特征提取模块提取的特征。
* Comp4-What-What特征提取，提取的深度学习特征。

```python
## 获得视频教程
from onekey_algo.custom.Manager import onekey_show
onekey_show('ICC')
```

```python
import pandas as pd

data1 = pd.read_csv('data/ICC1.csv', header=0)
data2 = pd.read_csv('data/ICC2.csv', header=0)

data1.head()
```

```python
import json
from onekey_algo.custom.components.comp1 import icc_filter_features

sel_features, features_icc = icc_filter_features([data1, data2], with_icc=True, threshold=0.75)
print(len(sel_features))
print(json.dumps(features_icc, indent=True, ensure_ascii=False))
```

```python
data1[sel_features].to_csv('data1_sel.csv', header=True, index=False)
data2[sel_features].to_csv('data2_sel.csv', header=True, index=False)
```
